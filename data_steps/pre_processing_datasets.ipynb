```python
data = pd.read_pickle("/content/df_event_data_with_embeddings.pkl")
```

```python
data
```

```python
dataset = data[["gid","text","predict","country_lat","country_lng","embeddings"]].copy()
```

```python
dataset
```

```python
dates = [date[:4]+'-'+date[4:6]+'-'+date[6:8] for date in dataset.gid]
```

```python
dataset['dates'] = pd.to_datetime(dates)
```

```python
dataset['country_lat'] = pd.to_numeric(dataset['country_lat'])
dataset['country_lng'] = pd.to_numeric(dataset['country_lng'])
```

```python
dataset.info()
```

```python
dataset = dataset.dropna(subset={'country_lat','country_lng'})
```

```python
dataset.info()
```

```python
dataset
```

```python
dataset['predict'].value_counts()
```

```python
labels = [label for label in dataset['predict'].unique() if dataset['predict'].value_counts()[label] > 1000]
```

```python
labels
```

```python
def formatting(str: str):
  return str.title().replace(" ","")
```

```python
roof = 10000
# floor = 1000
rstate = 42

for label in labels:
  label_data = dataset[dataset['predict']==label]

  if label_data.shape[0]>roof:
    df = label_data.sample(roof, random_state=rstate)
  else:
    # df = label_data.sample(floor, random_state=rstate)
    df = label_data
  print(df.shape)
  df.to_pickle(f"{formatting(label)}_dataset.pkl")
```
