In the following cells, we load the Dataset into a Pandas DataFrame. One may want to inspect the dataset to better understand our data in the associated processes. Remember events come from a news article.

```python
df_event_data = pd.read_pickle("df_event_data_with_embeddings.pkl") 
```

Next, we select the relevant features for our analysis, focusing on the semantic, geospatial, and temporal characteristics of each event. The selected features are:

* _gid_ – Unformatted date of the event news **(yyyy-mm-dd-hh-mm)**.
* _predict_ – A label that will be useful for sampling.

```python
df_selected_features = df_event_data[["gid","predict","country_lat","country_lng","embeddings"]].copy(deep=True)
```

# Data Cleaning & Feature Engineering 

Here, we check the data types and non-null counts of each relevant feature to guide the data cleaning and feature engineering process.

```python
df_selected_features.info()
```

The latitude and longitude coordinates are identified as `object`. We need to convert them to float type to enable future operations. 

```python
df_selected_features.country_lat.head(15)
```

One can see that there are null values, assigned with ''. Fortunately, `pd.to_numeric` handles this gracefully.

```python
df_selected_features['country_lat'] = pd.to_numeric(df_selected_features['country_lat'], errors='raise')
df_selected_features['country_lng'] = pd.to_numeric(df_selected_features['country_lng'], errors='raise')
```

```python
df_selected_features.info()
```

One can easily see 55,910 instances with null coordinates. A more in-depth discussion on handling these missing values is beyond our scope here. One may notice we'll sample the dataset, as a matter of fact.

We drop the data without geoespatial coordinates.

```python
df_selected_features = df_selected_features.dropna(subset={'country_lat','country_lng'})
```

Now, we need to get the temporal information we have about the events. This info is encoded gid and can be extracted as follows.


```python
dates = [date[:4]+'-'+date[4:6]+'-'+date[6:8] + ' ' + date[8:10] + ':' + date[10:12]  for date in df_selected_features.gid]

df_selected_features.loc[:, 'dates'] = pd.to_datetime(dates, errors='raise')
```

```python
df_selected_features
```

The resultant dataset got what we need to perform our experiments.

```python
df_processed = df_selected_features[['predict', 'country_lat', 'country_lng', 'embeddings','dates']]
```

```python
df_processed
```

# Creating Final Datasets

In this section, we present the code employed to sample the processed original data and create datasets based on the original labels. Preserving the labels allows us to establish a criterion for sampling the data.

### Utility functions

We define two functions: one to extract and format the label for creating sampled datasets, and another to generate medium-sized datasets from these samples.

```python
def formatting_label_name(str: str) -> str:
  return str.title().replace(" ","")

def create_datasets(data: pd.DataFrame, label: str, num_rows: int, random_state: int = 42) -> pd.DataFrame:
    dataset = data[data['predict'] == label]
    return dataset.sample(num_rows, random_state=random_state) if len(dataset) > num_rows else dataset
```
